# Phase4_Generative_AI

## – Prompt Design 

In this section, I prepared two different prompt designs for the email-classification task.  
The idea is to compare them and see which one gives clearer, more reliable responses for detecting “Spam” or “Not Spam”.  
Each prompt follows a simple structure and focuses on the main features we used in our ML model.


###  Prompt 1 

You are a *Spam Detection Assistant*.  
Your job is to classify an email as “Spam” or “Not Spam”.

Analyze the email using the following features:
- Suspicious keywords or phrases  
- Number of links or URLs  
- Urgent or threatening tone  
- Capital letter ratio / unusual formatting  
- Sender behavior (unknown sender, generic titles, rewards, offers)  
- Emotional manipulation (fear, urgency, free rewards)  
- Grammar issues or unnatural writing  
- Any patterns commonly found in phishing or spam emails  

After analyzing, provide the output in this exact format:

1. *Feature Analysis:*  
   - List the features you found in the email.

2. *Final Classification:*  
   - Choose one: *"Spam"* or *"Not Spam"*.

3. *Short Justification:*  
   - One sentence explaining the reason behind the classification.

*Email:*  
[PASTE EMAIL HERE]


###  Prompt 2

As a language model trained to understand email intent,
classify this email as “spam” or “not spam” based on meaning, tone, and structure.

Focus on:
- Fear, urgency, or pressure
- Links or requests for personal info
- Suspicious sender
- Unusual writing style

Output:
1. Predicted label: “spam” or “not spam”
2. Explanation: brief reasoning

Email: [PASTE EMAIL HERE]
