## Prompt Engineering

Prompt engineering is the process of **crafting prompts for a language model** to guide it in producing the most accurate, relevant, or desired output. Rather than modifying the underlying model, you effectively **“fine-tune”** its behavior through carefully designed prompts.


<img src="GenAi.png" width="600">

This is exactly what we did by leveraging our connection to the **powerful LLM from OpenAI**, enabling us to generate precise and context-aware responses tailored to our specific tasks.




---

## Prompt Design

In this section, we prepared **two different prompt designs** for the email-classification task.
The goal is to **compare their effectiveness** and determine which prompt produces **clearer and more reliable responses** for detecting “Spam” versus “Not Spam.”

Each prompt follows a **simple and structured format**, focusing on the main features used in our ML model, ensuring consistency and interpretability in the outputs.

---



###  Prompt 1 
You are a Spam Detection Assistant.  
Your job is to classify an email as "Spam" or "Not Spam".

Analyze the email using the following features:
•⁠  ⁠Suspicious keywords or phrases
•⁠  ⁠Number of links or URLs
•⁠  ⁠Urgent or threatening tone
•⁠  ⁠Capital letter ratio or unusual formatting
•⁠  ⁠Sender behavior (unknown sender, generic titles, rewards, offers)
•⁠  ⁠Emotional manipulation (fear, urgency, free rewards)
•⁠  ⁠Grammar issues or unnatural writing
•⁠  ⁠Any patterns commonly found in phishing or spam emails

After analyzing, provide the output in this exact format:

1.⁠ ⁠Feature analysis:
   - List the main features you found in the email.

2.⁠ ⁠Final classification:
   - Choose one → "Spam" or "Not Spam".

3.⁠ ⁠Short justification:
   - One sentence explaining why you classified it that way.

4.⁠ ⁠Safety advice:
   - If it is spam, give one practical tip to avoid this type of email.
   - If it is not spam, give one general email-safety reminder.

## 1.Connecting open Ai Api key

```python
from dotenv import load_dotenv
import os
from openai import OpenAI

# Load .env file (contains OPENAI_API_KEY)
load_dotenv(override=True)

api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    raise ValueError("OPENAI_API_KEY is not set in .env")

# Create OpenAI client
client = OpenAI(api_key=api_key)

print("API key loaded successfully.")
```

## 2. Send Request to Open Ai

```python
def get_gpt_advice(email_text: str, prompt_template: str) -> str:
    """
    Takes an email body and a prompt template,
    sends them to the GPT model, and returns the text response.
    """
    prompt = prompt_template.format(email_text=email_text)

    response = client.chat.completions.create(
        model="gpt-4.1-mini",      # cheap, reasoning-capable model
        messages=[
            {"role": "user", "content": prompt}
        ],
        max_tokens=400,
        temperature=0.4,
    )

    return response.choices[0].message.content.strip()
```

```python
PROMPT_1 = """
You are a Spam Detection Assistant.
Your job is to classify an email as "Spam" or "Not Spam".

Analyze the email using the following features:
•⁠  ⁠Suspicious keywords or phrases
•⁠  ⁠Number of links or URLs
•⁠  ⁠Urgent or threatening tone
•⁠  ⁠Capital letter ratio or unusual formatting
•⁠  ⁠Sender behavior (unknown sender, generic titles, rewards, offers)
•⁠  ⁠Emotional manipulation (fear, urgency, free rewards)
•⁠  ⁠Grammar issues or unnatural writing
•⁠  ⁠Any patterns commonly found in phishing or spam emails

After analyzing, provide the output in this exact format:

1.⁠ ⁠Feature analysis:
   - List the main features you found in the email.

2.⁠ ⁠Final classification:
   - Choose one ()"Spam" or "Not Spam").

3.⁠ ⁠Short justification:
   - One sentence explaining the reason.

4.⁠ ⁠Safety advice:
   - If it is spam, give one practical tip to avoid this type of email.
   - If it is not spam, give one general email-safety reminder.

Email:
{email_text}
""".strip()
```

```python
PROMPT_2 = """
Classify the following email as "spam" or "not spam" based mainly on its overall meaning, tone, and writing style.

Focus on:
•⁠  ⁠Fear, urgency, or pressure
•⁠  ⁠Links or requests for personal information
•⁠  ⁠Suspicious sender
•⁠  ⁠Unusual writing style

Output:
1.⁠ ⁠Predicted label: "spam" or "not spam"
2.⁠ ⁠Explanation: brief reasoning
3.⁠ ⁠Safety advice: one short tip depending on the classification

Email:
{email_text}
""".strip()
```

```python
test_email = """
Subject: Aramex Delivery – Action Required to Release Your Package

Dear Customer,

Your package is waiting for delivery, but we were unable to complete the shipment due to missing payment details.
To avoid returning the parcel, please confirm your information and pay the pending service fee (4.75 USD) within the next 24 hours.

Click the secure link below to proceed:
https://aramex-delivery-update-confirm.com/verify

Failure to confirm your details will result in package cancellation.

Thank you,
Aramex Support Team
""".strip()
```

```python
response_prompt1 = get_gpt_advice(test_email, PROMPT_1)
response_prompt2 = get_gpt_advice(test_email, PROMPT_2)

print("=== Prompt 1 Output ===\n")
print(response_prompt1)

print("\n\n" + "=" * 30 + "\n")

print("=== Prompt 2 Output ===\n")
print(response_prompt2)
```

## Comparison Between Prompt 1 and Prompt 2

Both prompts classified the test email as **spam**, and they focused on the same main signals such as the suspicious link, the urgent tone, and the request for payment.  
However, the style and level of detail in each prompt are different.

#### **Prompt 1**
- Gives a more **detailed and structured** analysis.  
- Breaks the email into multiple parts (keywords, tone, links, sender behavior, emotional cues, formatting, and common phishing patterns).
- Shows clearly *why* the email should be considered spam.  
- Matches the way we studied email patterns earlier, since it analyzes the message step-by-step.
- The advice generated from this prompt is usually more specific and actionable, because the model already analyzes many features in depth.
- Works best when we want a complete explanation and helpful guidance for the user.


#### **Prompt 2**
- Shorter and more **direct**.  
- Focuses only on a few key signals like urgency, suspicious link, and sender identity.  
- Produces a quick answer with a short explanation.  
- The advice produced here tends to be general because the prompt does not offer much structure for the model to think through.

#### **Overall**
- **Prompt 1**  is better when we want a more complete explanation, a clearer breakdown of the email, and stronger, more meaningful advice for the user.
- **Prompt 2** is useful for quick classification with minimal details, but its advice is less rich.
- Both prompts work, but Prompt 1 provides deeper insight and more helpful user guidance, making it more reliable for our system.
